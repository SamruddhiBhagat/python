```python
# Literals raw data given in variables
# 1. Numeric Literal
# 2. String Literal
# 3. Bollean Literal
# 4. Special Literal
```

```python
# 1. Numeric Literal
a = 0b1010
print("Binary =", a)

b = 100
print("Decimal =", b)

c = 0o310
print("Octal =", c)

d = 0xa
print("Hexadecimal =", d)


# float
print()
float_1 = 10.6
print("--- For float ---",)
print(float_1)

float_2 = 1.5e5
print(float_2)

float_3 = 1.5e-3
print(float_3)


print()
print("--- For complex ---",)
x = 1.4j
print(x,x.imag,x.real,sep = "/")
```

```python
# 2. String
str_1 = 'hello students'
print(str_1)

str_2 = "hello world"
print(str_2)

str_3 = """ welcome to our college, this is department of information technology"""
print(str_3)

unicode_str = u"\U0001f600"
print(unicode_str)
```

```python
# 3. Boolean
a = True
print(a)
result_1 = a + 4
print(result_1)
result_2 = True + 4
print(result_2)

print()

b = False
print(b)
result_3 = b + 10
print(result_3)
result_4 = False + 4
print(result_4)
```

```python
# 4. Special Literal
z = None
print(z)
```
