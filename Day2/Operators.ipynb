```python
# 1. Arithmetic operator
x = 10
y = 2

print("Addition", x+y)
print("Subtraction", x-y)
print("Multiplication", x*y)
print("Division (True Division)", x/y)
print("Integer Division", x//y)
print("Power", x**y)
```

```python
# 2. comparison operators
print(x>y)
print(x<y)
print(x==y)
print(x>=y)
print(x<=y)
print(x!=y)
```

```python
# 3. Logical Operator
a = True
b = False
print(a and b)
print(a or b)
print(not b)
```

```python
# 4. Bitwise operator
x = 2
y = 3
print(x & y)
print(x | y)
print(~x)
print(x>>2)
print(x<<3)
```

```python
# 5. Assignment operator
print("s =", s)
s = 10
s += 3
print("+=", s)

print()
print("s=", s)
s -= 3
print("-=", s)

print()
print("s=", s)
s *= 3
print("*=", s)

print()
print("s=", s)
s /= 3
print("/=", s)

print()
print("s=", s)
s %= 3
print("%=", s)
```

```python
# 6. Identity operator
a = "hello"
b = "hello"
print(a is b)
```

```python
a = 3
b = 3
print(a is b)
```

```python
a = [1,2,3]
b = [1,2,3]
print(a is b)
```

```python
# 7. Membership operator
# this is applicable for all like sets, dictionary , tuple, list, etc
a = "India"
print("I" in a)
```

```python
a = "India"
print("I" not in a)
```

```python
x = [1,2,3,4,5]
print(3 in x)
```
