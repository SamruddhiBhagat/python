```python
jackpot = random.randint(1,100)
guess = int(input("Enter the number = "))
counter = 1
while guess != jackpot :
        if guess < jackpot :
            print("Guess higher number")
        else :
            print("Guess lower number")
        guess = int(input("Enter the number = "))
        counter += 1
print("Correct guess")
print("You will take ", counter , "attempts")
```
