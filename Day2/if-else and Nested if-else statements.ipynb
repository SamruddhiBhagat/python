```python
mail = input("Enter email address = ")
if "@" in mail :
    password = input("Enter password = " )
    
    if mail == "samruddhi@gmail.com" and password == "1234" :
       print("Login successful")
    elif mail == "samruddhi@gmail.com" and password != "1234" :
        password = input("Again enter password = ")
        if password == "1234" :
            print("Finally the login successful")
        else :
            print("Still password incorrecct")
    else :
       print("Login failed")
else :
    print("Enter correct email address")
```
